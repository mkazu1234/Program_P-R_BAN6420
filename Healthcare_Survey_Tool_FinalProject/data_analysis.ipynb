```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

```

```python
# Load the CSV file
df = pd.read_csv('user_data.csv')

# Print column names to debug
print("Columns:", df.columns)

# Rename columns if necessary
df.rename(columns={
    'total_income': 'Total Income',  # Rename column for consistency
    'age': 'Age',                    # Rename column for consistency
    'gender': 'Gender',              # Ensure 'Gender' column exists
    'expenses.shopping': 'Shopping', # Rename 'expenses.shopping' to 'Shopping'
    'expenses.utilities': 'Utilities',
    'expenses.entertainment': 'Entertainment',
    'expenses.school_fees': 'School Fees',
    'expenses.healthcare': 'Healthcare'
}, inplace=True)

# Convert to numeric if necessary
df['Total Income'] = pd.to_numeric(df['Total Income'], errors='coerce')
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')

# Handle missing values
df.dropna(subset=['Total Income', 'Age'], inplace=True)

# Show ages with the highest income
plt.figure(figsize=(12, 6))
sns.scatterplot(data=df, x='Age', y='Total Income', alpha=0.6)
plt.title('Age vs Total Income')
plt.xlabel('Age')
plt.ylabel('Total Income')
plt.savefig('age_vs_income.png')
plt.close()

# Show gender distribution across spending categories
expense_categories = ['Utilities', 'Entertainment', 'School Fees', 'Shopping', 'Healthcare']
melted_df = pd.melt(df, id_vars=['Gender'], value_vars=expense_categories, var_name='Category', value_name='Amount')

plt.figure(figsize=(12, 6))
sns.boxplot(data=melted_df, x='Category', y='Amount', hue='Gender')
plt.title('Gender Distribution Across Spending Categories')
plt.xlabel('Expense Category')
plt.ylabel('Amount Spent')
plt.savefig('gender_distribution.png')
plt.close()

print("Charts have been exported as 'age_vs_income.png' and 'gender_distribution.png'")


```
